```python
29.
import pandas as pd

# Create a labelled Series
sales = pd.Series(
    [300, 650, 450, 800],
    index=['Pen', 'Notebook', 'Bag', 'Laptop']
)

# Display values
print("Values:")
print(sales.values)

# Display index
print("\nIndex:")
print(sales.index)

# Select one label
print("\nNotebook sales:")
print(sales['Notebook'])

# Filter sales above 500
print("\nSales above 500:")
print(sales[sales > 500])

# Check whether a label exists
print("\nIs Bag present?")
print('Bag' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()

print("\nDictionary:")
print(sales_dict)
```
